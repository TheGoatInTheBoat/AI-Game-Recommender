```python
API_KEY = "C45399540985DC199A1C374C8EF89F6F"

def VerifyInput(query_set):
    if len(query_set) != 5:
        print("Game count not valid!!!")
        return False
    for game in query_set:
        id = game["ID"]
        url = f"https://store.steampowered.com/api/appdetails?appids={id}&cc=us"
        try:
            data = requests.get(url).json()
        except (MissingSchema, InvalidURL) as url_err:
            print(f"Game {id} is not available on store page")
            return False
        if data[f"{id}"]["success"] == False:
            print(f"Game {id} is not available on store page")
            return False
    for i, game_i in enumerate(query_set):
        for j, game_j in enumerate(query_set):
            if i != j and game_i["ID"] == game_j["ID"]:
                print(f"Repeated input found!!!")
                return False
    return True
```
